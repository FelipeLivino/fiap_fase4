# 10 - Prototipo de Inferencia em Notebook

Este notebook implementa a tarefa 23 da ordem recomendada: criar um prototipo simples de inferencia.

Ele carrega o modelo final exportado em `models/exported/`, recebe uma imagem de raio-X e retorna:

1. Classe prevista.
2. Probabilidade de `Cardiomegaly`.
3. Probabilidade de `No Finding`.
4. Aviso de uso exclusivamente academico.

**Aviso:** este prototipo nao e uma ferramenta diagnostica real.

## Pre-requisitos

Execute antes os notebooks de treinamento e depois:

`notebooks/09_comparacao_modelos.ipynb`

Esse notebook exporta o checkpoint escolhido para `models/exported/`.

```python
from __future__ import annotations

import sys
from pathlib import Path

import matplotlib.pyplot as plt
from PIL import Image

CURRENT_DIR = Path.cwd().resolve()
PROJECT_ROOT = CURRENT_DIR if (CURRENT_DIR / "src").exists() else CURRENT_DIR.parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src import config
from src.inference import load_model_from_checkpoint, predict_image

config.ensure_project_directories()

print("DEVICE:", config.DEVICE)
print("EXPORTED_MODELS_DIR:", config.EXPORTED_MODELS_DIR)
```

## Carregar modelo final

Se ainda nao houver checkpoint exportado, esta celula indica qual etapa deve ser executada.

```python
model, model_metadata = load_model_from_checkpoint(device=config.DEVICE)
model_metadata
```

## Selecionar imagem

Defina `IMAGE_PATH` com o caminho de uma imagem local. Pode ser uma imagem de teste usada no dataset.

```python
IMAGE_PATH = None  # exemplo: "data/raw/images/00000001_000.png"

if IMAGE_PATH is None:
    raise ValueError("Defina IMAGE_PATH com o caminho de uma imagem para executar a inferencia.")

image_path = Path(IMAGE_PATH)
if not image_path.exists():
    raise FileNotFoundError(f"Imagem nao encontrada: {image_path}")

prediction = predict_image(
    image_path=image_path,
    model=model,
    model_name=model_metadata["model_name"],
    device=config.DEVICE,
)
prediction
```

## Resultado visual

```python
image = Image.open(image_path).convert("RGB")

plt.figure(figsize=(7, 7))
plt.imshow(image, cmap="gray")
plt.axis("off")
plt.title(
    f"Predicao: {prediction.predicted_label}\n"
    f"P(Cardiomegaly): {prediction.probability_cardiomegaly:.2%} | "
    f"P(No Finding): {prediction.probability_no_finding:.2%}\n"
    f"Modelo: {prediction.model_name}"
)
plt.show()

print("Este prototipo possui finalidade exclusivamente academica e educacional.")
print("Ele nao deve ser usado como ferramenta diagnostica real.")
```
